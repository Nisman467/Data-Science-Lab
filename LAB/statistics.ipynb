```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
from scipy import stats

#House Price
house_price = [10000000, 25000000, 12000000, 25000000, 13000000, 24000000, 25000000]

#Calulation 
mean_house_price = np.mean(house_price)
median_house_price = np.median(house_price)
mode_house_price = stats.mode(house_price, keepdims=True)

#Display output
print(f"Mean from the house price = ${mean_house_price:,.2f}")
print(f"Median from the house price = ${median_house_price:,.2f}")
print(f"Mode from the house price =  ${mode_house_price.mode[0]:,.2f}\tCount: {mode_house_price.count[0]}")
```

```python
#House Price
test_scores = [56, 65, 67, 70, 72, 75, 80, 82, 85, 90, 95, 100]

#Calulation 
mean_test_score = np.mean(test_scores)
median_test_score = np.median(test_scores)
mode_test_score = stats.mode(test_scores, keepdims=True)

#Display output
print(f"Mean from the house price = ${mean_test_score:,.2f}")
print(f"Median from the house price = ${median_test_score:,.2f}")
print(f"Mode from the house price =  ${mode_test_score.mode[0]:,.2f}\tCount: {mode_test_score.count[0]}")
```

```python
#The dataset of friends in facebook of every individual 
friends = [109, 1017, 1127, 418, 625, 957, 89, 950, 946, 797, 981,
           125, 455, 731, 1640, 485, 1309, 472, 1132, 1773, 906, 531, 742, 621]

#Calculation 
mean_friends = np.mean(friends)
median_friends = np.median(friends)
standard_deviation_friends = np.std(friends) #std means standard deivation 
range_friends = np.max(friends) - np.min(friends)

#Displaying the outputs
print(f"Mean from friends = {mean_friends:,.2f}")
print(f"Median from friends = {median_friends:,.2f}")
print(f"Range of friends dataset = {range_friends}")
print(f"Standard deviation of friends dataset = {standard_deviation_friends:,.2f}")
```

Graphical representation of the friends datasets where the (mean + standard deviation) is denoted as green, 
(mean - standard deviation) denoted as blue and center points by red.

#Finding the quartiles 

```python
#using the same friend dataset
Q1 = np.percentile(friends, 25)
Q3 = np.percentile(friends, 75)
IQR = Q3 - Q1

#Calculate lower and upper boundary
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

#Removing the outliers
filtered_friends = [i for i in friends if lower_bound <= i <= upper_bound]

#Ploting the before and after boxplots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize = (12, 6))

#Before removing the outliers
ax1.boxplot(friends)
ax1.set_title("Before Removing Outliers")

#After removing the outliers
ax2.boxplot(filtered_friends)
ax2.set_title("After Removing Outliers")

plt.tight_layout()
plt.show()
```

```python
#dataset 
Scores=[56,78,85,92,61,74,88,93,67,81,130]

#using the same friend dataset
Q1 = np.percentile(Scores, 25)
Q3 = np.percentile(Scores, 75)
IQR = Q3 - Q1

#Calculate lower and upper boundary
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

#Removing the outliers
filtered_Scores = [i for i in Scores if lower_bound <= i <= upper_bound]

#Ploting the before and after boxplots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize = (12, 6))

#Before removing the outliers
ax1.boxplot(Scores)
ax1.set_title("Before Removing Outliers")

#After removing the outliers
ax2.boxplot(filtered_Scores)
ax2.set_title("After Removing Outliers")

plt.tight_layout()
plt.show()
```

```python
z_scores = []

m_friends = np.mean(friends)
std_friends = np.std(friends)

for x in friends:
    z = (x - m_friends)/ std_friends
    z_scores.append(z)

plt.bar(y_pos, z_scores)

```

```python
from sklearn import preprocessing

happiness = [.8, .6, .3, .6, .6, .4, .8, .5, .4, .3, .3, .6, .2, .8,
1, .6, .2, .7, .5, .3, .1, 0, .3, 1]

df = pd.DataFrame({'friends': friends, 'happiness': happiness})
print(df.head())

## Steps for z score
# Finding the mean of the column
# Finding the standard deviation of the column
# Applying the z-score function to each element in the column
df_scaled = pd.DataFrame(preprocessing.scale(df), columns=['friends_scaled', 'happiness_scaled'])
df_scaled.head()

df_scaled.plot(kind='scatter', x = 'friends_scaled', y = 'happiness_scaled')
```

```python
df.corr()
```

```python
# finding the percentage of people within one standard deviation of the mean
within_1_std = df_scaled[(df_scaled['friends_scaled'] <= 1) & (df_scaled['friends_scaled'] >= -1)].shape[0]
print("within 1 std",within_1_std / float(df_scaled.shape[0]))
# 0.75
# finding the percentage of people within two standard deviations of the mean
within_2_std = df_scaled[(df_scaled['friends_scaled'] <= 2) & (df_scaled['friends_scaled'] >= -2)].shape[0]
print("within 2 std",within_2_std / float(df_scaled.shape[0]))
# 0.916
# finding the percentage of people within three standard deviations of the mean
within_3_std = df_scaled[(df_scaled['friends_scaled'] <= 3) & (df_scaled['friends_scaled'] >= -3)].shape[0]
print("within 3std",within_3_std / float(df_scaled.shape[0]))
```
