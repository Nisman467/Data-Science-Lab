```python
# 3. Add Intercept Column (ones)
m_train = X_train_scaled.shape[0]
m_test = X_test_scaled.shape[0]
X_train_final = np.concatenate((np.ones((m_train, 1)), X_train_scaled), axis=1)
X_test_final = np.concatenate((np.ones((m_test, 1)), X_test_scaled), axis=1)
```

```python
# Now finding the gradient

theta = np.zeros(X_train_final.shape[1])
lr = 0.001  # Lower learning rate to be safe
max_iter = 2000 

for i in range(max_iter):
    y_hat = X_train_final @ theta
    # Subtracting two numpy arrays is safe
    error = y_hat - y_train
    grad = (1/m_train) * (X_train_final.T @ error)
    theta = theta - lr * grad
    
    if i % 500 == 0:
        mse = np.mean(error**2)
        print(f"Iteration {i}: MSE = {mse:.4f}")

yhat_test = X_test_final @ theta

print(f"Prediction Max: {yhat_test.max():.2f}, Min: {yhat_test.min():.2f}")
```
